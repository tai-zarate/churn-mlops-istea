## Hallazgo 1: valores faltantes en TotalCharges

26 valores nulos, todos en clientes con tenure = 0.
Hay 30 clientes con tenure = 0: 26 tienen TotalCharges nulo y 4 tienen 0.0 explícito.
El nulo no es un error de carga, representa "todavía no facturó".

**Decisión para el pipeline:** imputar TotalCharges = 0 cuando falta.
No usar media ni mediana: asignaría ~1400 de historial de facturación
a clientes que recién ingresaron, justo el segmento con mayor churn.

## Hallazgo 2: categorías redundantes en servicios de internet

1549 clientes (22%) tienen InternetService = "No".
Ese mismo grupo aparece como "No internet service" en 6 columnas:
OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport,
StreamingTV, StreamingMovies.

Al aplicar OneHotEncoder se generarían 6 variables binarias idénticas
entre sí e idénticas a InternetService_No: colinealidad perfecta,
coeficientes de la regresión logística no interpretables.

**Decisión para el pipeline:** recodificar "No internet service" como "No"
en las 6 columnas. InternetService queda como única variable que informa
si el cliente tiene internet.

## Hallazgo 3: desbalance de clases

Churn = No: 73.6%

Churn = Yes: 26.4%

Un clasificador trivial que prediga siempre "No" alcanza 73.6% de accuracy
sin aprender nada. La accuracy no discrimina un modelo útil de uno inútil
en este dataset.

**Decisión para el pipeline:** métrica principal recall sobre la clase
positiva, acompañada de precision, F1 y ROC-AUC. Evaluar class_weight='balanced'
en los modelos y ajustar el umbral de decisión.

**Justificación de negocio:** el falso negativo (no detectar a quien se va)
cuesta el cliente completo más el costo de readquisición. El falso positivo
(retener a quien se quedaba) cuesta solo el margen del descuento.
La asimetría favorece recall sobre precision.
